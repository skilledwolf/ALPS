##  Two-dimensional Heisenberg square lattice  

The Python cells below create TOML input files. Validate one with `dirloop_sse --validate parm4.task1.toml`, then run it with `dirloop_sse parm4.task1.toml`.


### Preparing and running the simulation using Python 

The following cells write a TOML job and run the solver. Results use native HDF5 files.

```python
import pyalps

parms = [{ 
         'LATTICE'                   : "square lattice", 
         'MODEL'                     : "spin",
         'MEASURE[Correlations]'     : True,
         'MEASURE[Structure Factor]' : True,
         'MEASURE[Green Function]'   : True,
         'local_S'                   : 0.5,
         'T'                         : 0.3,
         'J'                         : 1 ,
         'THERMALIZATION'            : 10000,
         'SWEEPS'                    : 500000,
         'L'                         : 4,
         'h'                         : 0.1
       }]
from pyalps.run_io import execute, write_run_files
runs = [dict(parameters=p, output=dict(results=f'parm4.task{i+1}.out.h5'))
        for i, p in enumerate(parms)]
files = execute('dirloop_sse', write_run_files('parm4', runs, baseseed=42, overwrite=True))
```

The following cells write a TOML job and run the solver. Results use native HDF5 files.

### Evaluating the simulation and preparing plots using Python 

To look at the results we now load ALL measurements from output files starting with <tt>parm4</tt>. The script is again in [tutorial4.py](../../03-mc/04-measurements/tutorial4.py)

```python
data = pyalps.loadMeasurements(files)
```

We now loop through all measurements and print them:

```python
for s in pyalps.flatten(data):
    if len(s.x)==1:
        print(s.props['observable'], ' : ', s.y[0])
    else:
        for (x,y) in zip(s.x,s.y):
            print(s.props['observable'], x, ' : ', y)
```

The if statement checks whether the measured quantity is a scalar or vector-valued quantity.
